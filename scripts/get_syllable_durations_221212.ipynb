### Loop attempt 1
(1) Goes through all the TextGrids in the folder
(2) Saves timestamps for each syllable interval
(3) Calculates vowel durations based on those timestamps

```python
def get_grids():
    grids = glob('test_voices/extras/*.auto.TextGrid')
    return grids
```

```python
dur_list = [[],[]] # Two-dimensional empty list

#id_list = []
if __name__ == "__main__":
    grids = get_grids()

    for grid in grids:
        tg = textgrid.openTextgrid(grid, includeEmptyIntervals=False)
        tier = tg.tierDict[tg.tierNameList[2]]

        start_list =[]
        stop_list = []
        label_list = []
    
        for start, stop, label in tier.entryList:
            start_list.append(start)
            stop_list.append(stop)
            label_list.append(label)
        
        # Default 12:15
        # Japanese 21:24
        # Polish 19-22
        # Japanese+4 25:28
        with open('sylldur_' + str(grid[19:22]) + '.csv', 'w', newline = '') as f:
            write = csv.writer(f)
            for i in range(len(start_list)):
                write.writerow([start_list[i], stop_list[i], label_list[i]])
                df = pd.DataFrame(np.column_stack([start_list, stop_list, label_list, np.subtract(stop_list, start_list)]),
                                  columns=['start', 'stop', 'type', 'diff'])
                df.to_csv('sylldur_' + str(grid[19:22]) + '_added.csv', index=False)
                    
            fileid = str(grid[19:22]) # Get file ID
            diff = df['diff']
            diff = pd.to_numeric(diff)
            dursd = diff.std()
            dur_list[0].append(dursd) # Add SD duration
            dur_list[1].append(fileid) # Add file ID
              
        #durdf = pd.DataFrame(dur_list)
        durdf = pd.DataFrame({'id': dur_list[1], 'dur': dur_list[0]})
        durdf.to_csv('sylldurs_extras.csv', index=False)
```
